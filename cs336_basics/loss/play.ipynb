```python
import torch
B = 4
T = 4
vocab_size = 64
logits = torch.randn((B,T,vocab_size))
targets = torch.randint(size = (B,T,1), low = 0, high = vocab_size-1)
```

```python
result = torch.gather(logits, dim = -1, index = targets)
result.shape
```

```python
torch.sum(torch.exp(logits), dim = -1).shape
```

```python
result = - result + torch.sum(torch.exp(logits), dim = -1)
```

```python
result = torch.mean(result)
```
